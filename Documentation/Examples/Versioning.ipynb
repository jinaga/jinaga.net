```polyglot-notebook
using Jinaga;
using Jinaga.UnitTest;
using Jinaga.Graphviz;
```

```polyglot-notebook
var jinagaClient = JinagaTest.Create();
```

```polyglot-notebook
[FactType("Blog.Site")]
internal record Site(string domain) {}

[FactType("Blog.Post")]
internal record Post(Site site, DateTime createdAt) {}
```

```polyglot-notebook
var site = await jinagaClient.Fact(new Site("example.com"));
var post = await jinagaClient.Fact(new Post(site, "Hello, world!"));

Renderer.RenderFacts(post)
```

```polyglot-notebook
var postsInSite = Given<Site>.Match((site, facts) =>
    from post in facts.OfType<Post>()
    where post.site == site
    select post
);

var posts = await jinagaClient.Query(postsInSite, new Site("example.com"));

posts.Select(p => p)
```
