# CoLab

Colaborative project management.

```polyglot-notebook
#r "nuget: Jinaga, 0.11.14"
#r "nuget: Jinaga.Graphviz, 0.11.14"

using Jinaga;
using Jinaga.Graphviz;
```

```polyglot-notebook
var j = JinagaClient.Create(opt =>
{
  opt.HttpEndpoint = new Uri("http://localhost:8080/jinaga");
});
```

A user creates a project.
The project has a name.

```polyglot-notebook
[FactType("Jinaga.User")]
public record User(string publicKey) {}

[FactType("CoLab.Project")]
public record Project(User creator, DateTime createdAt) {}

[FactType("CoLab.Project.Name")]
public record ProjectName(Project project, string value, ProjectName[] prior) {}

Renderer.RenderTypes(typeof(User), typeof(Project), typeof(ProjectName))
```

For example, a user creates a project called "Wingman".

```polyglot-notebook
var michael = await j.Fact(new User("---PUBLIC KEY---Michael---END PUBLIC KEY---"));
var wingman = await j.Fact(new Project(michael, DateTime.Parse("2024-04-07T15:00:00.000Z")));
ProjectName prior = await j.Fact(new ProjectName(wingman, $"Initial name", new ProjectName[0]));
List<ProjectName> concurrentNames = new List<ProjectName>();
for (var i = 0; i < 1000; i++)
{
  var name = new ProjectName(wingman, $"New name {i}", [ prior ]);
  concurrentNames.Add(name);
}

var resolution = await j.Fact(new ProjectName(wingman, $"Final name", concurrentNames.ToArray()));
```

```polyglot-notebook

j.RenderFacts(resolution)
```

Run a projection to list the projects.

```polyglot-notebook
var projectsFromUser = Given<User>.Match((user, facts) =>
  from project in facts.OfType<Project>()
  where project.creator == user
  select new {
    createdAt = project.createdAt,
    names =
      from name in facts.OfType<ProjectName>()
      where name.project == project &&
        !facts.Any<ProjectName>(next => next.prior.Contains(name))
      select name.value
  }
);

var projects = await j.Query(projectsFromUser, michael);

projects
```

The preceding specification includes a projection, which selects certain interesting fields of the facts.
To render the graph, we'll use a specification that selects the facts themselves.

```polyglot-notebook
var projectFactsFromUser = Given<User>.Match((user, facts) =>
  from project in facts.OfType<Project>()
  where project.creator == user
  select project
);

var projectNamesFromUser = Given<User>.Match((user, facts) =>
  from name in facts.OfType<ProjectName>()
  where name.project.creator == user &&
    !facts.Any<ProjectName>(next => next.prior.Contains(name))
  select name
);

var projectFacts = await j.Query(projectFactsFromUser, michael);
var projectNames = await j.Query(projectNamesFromUser, michael);

j.RenderFacts(projectFacts, projectNames)
```
